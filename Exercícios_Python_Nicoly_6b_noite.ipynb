## 21. Busca Binária

a busca binária só funciona em lista ordenada, ela divide a lista ao meio e vai descartando a metade que não interessa até achar o número

```python
# função que procura o alvo numa lista ordenada dividindo ela ao meio a cada tentativa
def busca_binaria(lista, alvo):
    inicio = 0
    fim = len(lista) - 1

    while inicio <= fim:
        meio = (inicio + fim) // 2
        if lista[meio] == alvo:
            return meio
        elif lista[meio] < alvo:
            inicio = meio + 1
        else:
            fim = meio - 1

    return -1

#lista já ordenada, obrigatorio pra busca binaria funcionar
numeros = [2, 5, 8, 12, 16, 23, 38, 45, 56, 72]
alvo = int(input("Digite o número que quer procurar: "))

posicao = busca_binaria(numeros, alvo)

if posicao != -1:
    print(f"Número encontrado na posição {posicao}")
else:
    print("Número não está na lista")
```

## 22. Criptografia de Cifra de César

a cifra de cesar desloca cada letra do alfabeto um numero fixo de posições, usei ord() e chr() pra converter entre letra e numero

```python
# função que desloca cada letra da mensagem um número fixo de posições no alfabeto
def cifrar_cesar(mensagem, deslocamento):
    resultado = ""

    for letra in mensagem:
        if letra.isalpha():
            base = ord('A') if letra.isupper() else ord('a')
            # desloca a letra e usa % 26 pra voltar pro inicio do alfabeto se passar do z
            nova_letra = chr((ord(letra) - base + deslocamento) % 26 + base)
            resultado += nova_letra
        else:
            resultado += letra

    return resultado

mensagem = input("Digite a mensagem: ")
deslocamento = int(input("Digite o deslocamento: "))

criptografada = cifrar_cesar(mensagem, deslocamento)
print("Mensagem criptografada:", criptografada)
```

## 23. Descriptografar Cifra de César

pra descriptografar é só usar a mesma função do exercicio anterior com o deslocamento negativo

```python
# função que desfaz a cifra de cesar aplicando o deslocamento ao contrário
def descifrar_cesar(mensagem, deslocamento):
    # descriptografar é a mesma coisa que cifrar só que com deslocamento negativo
    return cifrar_cesar(mensagem, -deslocamento)

mensagem_cifrada = input("Digite a mensagem criptografada: ")
deslocamento = int(input("Digite o deslocamento usado: "))

original = descifrar_cesar(mensagem_cifrada, deslocamento)
print("Mensagem original:", original)
```

## 24. Validação de CPF

cpf tem 2 digitos verificadores calculados com peso sobre os numeros anteriores, calculei os dois digitos e comparei com os informados

```python
# função que calcula os 2 dígitos verificadores do cpf e compara com os que foram digitados
def validar_cpf(cpf):
    cpf = cpf.replace(".", "").replace("-", "")

    if len(cpf) != 11 or cpf == cpf[0] * 11:
        return False

    # calcula o primeiro digito verificador
    soma = sum(int(cpf[i]) * (10 - i) for i in range(9))
    resto = (soma * 10) % 11
    digito1 = 0 if resto == 10 else resto

    # calcula o segundo digito verificador
    soma = sum(int(cpf[i]) * (11 - i) for i in range(10))
    resto = (soma * 10) % 11
    digito2 = 0 if resto == 10 else resto

    return cpf[-2:] == f"{digito1}{digito2}"

cpf = input("Digite o CPF (só números): ")

if validar_cpf(cpf):
    print("CPF válido")
else:
    print("CPF inválido")
```

## 30. Jogo da Forca

o jogador vai tentando letra por letra, cada erro perde uma vida, o jogo acaba quando acerta a palavra toda ou zera as vidas

```python
#palavra secreta e quantidade de vidas
palavra = "python"
letras_certas = ["_"] * len(palavra)
vidas = 6
letras_tentadas = []

#continua enquanto tiver vida e ainda tiver letra faltando
while vidas > 0 and "_" in letras_certas:
    print("Palavra:", " ".join(letras_certas))
    print("Vidas:", vidas)

    letra = input("Digite uma letra: ").lower()

    if letra in letras_tentadas:
        print("Você já tentou essa letra")
        continue

    letras_tentadas.append(letra)

    if letra in palavra:
        # atualiza as posições onde a letra aparece
        for i in range(len(palavra)):
            if palavra[i] == letra:
                letras_certas[i] = letra
    else:
        vidas -= 1
        print("Letra errada")

if "_" not in letras_certas:
    print("Você venceu! A palavra era:", palavra)
else:
    print("Você perdeu. A palavra era:", palavra)
```

## 31. Ordenação por Bolha

o bubble sort compara os numeros vizinhos e vai trocando de lugar quando estão na ordem errada, repete isso ate a lista ficar ordenada

```python
#lista pra ordenar
numeros = [9, 3, 7, 1, 8, 2]

n = len(numeros)

#percorre a lista varias vezes comparando vizinhos
for i in range(n):
    for j in range(0, n - i - 1):
        if numeros[j] > numeros[j + 1]:
            # troca os dois numeros de posição
            numeros[j], numeros[j + 1] = numeros[j + 1], numeros[j]

print("Lista ordenada:", numeros)
```

## 32. Jogo da Velha

tabuleiro representado como uma lista de 9 posições, os jogadores vão escolhendo a posição até alguem ganhar ou empatar

```python
tabuleiro = [" "] * 9

# função que imprime o tabuleiro no formato 3x3
def mostrar_tabuleiro():
    print(f"{tabuleiro[0]}|{tabuleiro[1]}|{tabuleiro[2]}")
    print(f"{tabuleiro[3]}|{tabuleiro[4]}|{tabuleiro[5]}")
    print(f"{tabuleiro[6]}|{tabuleiro[7]}|{tabuleiro[8]}")

# função que confere todas as combinações de vitória (linha, coluna, diagonal)
def verificar_vencedor():
    combinacoes = [(0,1,2), (3,4,5), (6,7,8), (0,3,6), (1,4,7), (2,5,8), (0,4,8), (2,4,6)]
    for a, b, c in combinacoes:
        if tabuleiro[a] == tabuleiro[b] == tabuleiro[c] != " ":
            return tabuleiro[a]
    return None

jogador_atual = "X"

for jogada in range(9):
    mostrar_tabuleiro()
    posicao = int(input(f"Jogador {jogador_atual}, escolha uma posição (0-8): "))

    if tabuleiro[posicao] == " ":
        tabuleiro[posicao] = jogador_atual
        vencedor = verificar_vencedor()
        if vencedor:
            mostrar_tabuleiro()
            print(f"Jogador {vencedor} venceu!")
            break
        jogador_atual = "O" if jogador_atual == "X" else "X"
    else:
        print("Posição já ocupada, tente de novo")
else:
    print("Empate!")
```

## 33. Verificação de Anagramas

mesma ideia do exercicio 27, mas aqui verifiquei contando as letras com um dicionario ao inves de usar sorted()

```python
# função que conta as letras das duas palavras num dicionário e compara se bateram
def sao_anagramas(p1, p2):
    p1 = p1.lower().replace(" ", "")
    p2 = p2.lower().replace(" ", "")

    if len(p1) != len(p2):
        return False

    contagem = {}

    for letra in p1:
        contagem[letra] = contagem.get(letra, 0) + 1

    for letra in p2:
        if letra not in contagem:
            return False
        contagem[letra] -= 1

    return all(valor == 0 for valor in contagem.values())

#recebe as 2 palavras do usuario
palavra1 = input("Digite a primeira palavra: ")
palavra2 = input("Digite a segunda palavra: ")

if sao_anagramas(palavra1, palavra2):
    print("São anagramas")
else:
    print("Não são anagramas")
```

## 37. Funções Recursivas

fiz o fatorial de forma recursiva, a função chama ela mesma até chegar no caso base que é quando o numero é 0 ou 1

```python
# função que calcula o fatorial chamando ela mesma até chegar no caso base
def fatorial_recursivo(n):
    # caso base, para a recursão
    if n == 0 or n == 1:
        return 1
    # chama a função de novo com n-1
    return n * fatorial_recursivo(n - 1)

numero = int(input("Digite um número: "))
print(f"O fatorial de {numero} é {fatorial_recursivo(numero)}")
```

## 42. Números Primos até N

```python
# fator de conversão de cada unidade para metros
fatores = {
    "km": 1000,
    "m": 1,
    "cm": 0.01,
    "mm": 0.001
}

valor = float(input("Digite o valor: "))
unidade_origem = input("Unidade de origem (km, m, cm, mm): ")
unidade_destino = input("Unidade de destino (km, m, cm, mm): ")

# converte pra metro primeiro e depois pra unidade final
metros = valor * fatores[unidade_origem]
resultado = metros / fatores[unidade_destino]

print(f"{valor} {unidade_origem} equivale a {resultado} {unidade_destino}")
```

## 44. Fibonacci em Nível

aqui o usuario escolhe quantos termos ele quer, diferente do 35 que era baseado num valor limite

```python
#recebe quantos termos o usuario quer
termos = int(input("Quantos termos você quer? "))

a, b = 0, 1
fibonacci = []

#gera a quantidade de termos pedida
for _ in range(termos):
    fibonacci.append(a)
    a, b = b, a + b

print("Sequência de Fibonacci:", fibonacci)
```

## 45. Par ou Ímpar em Lista

percorri a lista e classifiquei cada numero usando o resto da divisão por 2

```python
#lista de numeros
numeros = [4, 7, 10, 15, 22, 33, 8]

pares = []
impares = []

#classifica cada numero pelo resto da divisao por 2
for n in numeros:
    if n % 2 == 0:
        pares.append(n)
    else:
        impares.append(n)

print("Pares:", pares)
print("Ímpares:", impares)
```

## 46. Calcular Média e Desvio Padrão

a media é a soma dividido pela quantidade, o desvio padrão eu calculei pela formula, a raiz da media dos quadrados das diferenças pra media

```python
import math

#lista de numeros
numeros = [10, 12, 23, 23, 16, 23, 21, 16]

#calcula a media
media = sum(numeros) / len(numeros)

# soma o quadrado da diferença de cada numero pra media
soma_quadrados = sum((n - media) ** 2 for n in numeros)
desvio_padrao = math.sqrt(soma_quadrados / len(numeros))

print("Média:", media)
print("Desvio padrão:", desvio_padrao)
```

## 47. Contar Números em Lista

usei um dicionario pra guardar quantas vezes cada numero aparece na lista

```python
#lista de numeros
numeros = [1, 2, 2, 3, 3, 3, 4, 1, 5, 2]

#dicionario pra guardar quantas vezes cada numero aparece
contagem = {}

for n in numeros:
    contagem[n] = contagem.get(n, 0) + 1

print("Contagem de cada número:", contagem)
```

## 48. Ordenação por Inserção

o insertion sort pega cada numero e vai colocando ele na posição certa comparando com os numeros anteriores

```python
#lista pra ordenar
numeros = [7, 2, 9, 4, 1, 6]

#pega cada numero e vai colocando na posição certa
for i in range(1, len(numeros)):
    atual = numeros[i]
    j = i - 1

    # vai empurrando os numeros maiores pra frente ate achar o lugar certo
    while j >= 0 and numeros[j] > atual:
        numeros[j + 1] = numeros[j]
        j -= 1

    numeros[j + 1] = atual

print("Lista ordenada:", numeros)
```

## 49. Soma dos Dígitos

converti o numero pra string pra poder percorrer digito por digito e ir somando

```python
#recebe o numero como string pra poder percorrer digito por digito
numero = input("Digite um número inteiro: ")

soma = 0

#soma cada digito
for digito in numero:
    soma += int(digito)

print("Soma dos dígitos:", soma)
```

## 50. Cálculo de Frequência de Letras

parecido com o exercicio 12, mas aqui filtrei só as letras ignorando espaço e outros caracteres

```python
#recebe a string
texto = input("Digite uma string: ").lower()

frequencia = {}

#percorre so as letras, ignorando espaço e outros caracteres
for letra in texto:
    if letra.isalpha():
        frequencia[letra] = frequencia.get(letra, 0) + 1

print("Frequência de cada letra:", frequencia)
```
